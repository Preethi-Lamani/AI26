```python
import random

def calculate_conflicts(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                conflicts += 1

    return conflicts


def display_board(board):
    n = len(board)

    for row in range(n):
        for col in range(n):
            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


def hill_climbing(n):
    board = [random.randrange(n) for _ in range(n)]
    steps = 0

    while True:
        conflicts = calculate_conflicts(board)

        if conflicts == 0:
            print("\nSolution Found!")
            print("Final Board:")
            display_board(board)
            print("Number of attacking pairs:", conflicts)
            print("Number of steps taken:", steps)
            return

        best_board = board
        best_conflicts = conflicts

        # Generate neighboring states
        for col in range(n):
            for row in range(n):
                if row != board[col]:
                    neighbor = board.copy()
                    neighbor[col] = row

                    cost = calculate_conflicts(neighbor)

                    if cost < best_conflicts:
                        best_conflicts = cost
                        best_board = neighbor

        # Local minimum: restart randomly
        if best_conflicts >= conflicts:
            board = [random.randrange(n) for _ in range(n)]
            steps += 1
        else:
            board = best_board
            steps += 1


# Main program
n = int(input("Enter the number of queens: "))

if n < 1:
    print("Invalid input! N must be at least 1.")
elif n == 2 or n == 3:
    print("Solution does not exist for N =", n)
else:
    print("\nInitial Board:")
    initial_board = [random.randrange(n) for _ in range(n)]
    display_board(initial_board)
    print("Number of attacking pairs:",
          calculate_conflicts(initial_board))

    print("\nApplying Hill Climbing...")
    hill_climbing(n)
```
