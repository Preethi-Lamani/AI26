```python
board = [" "] * 9

def display():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])


def win(p):
    combinations = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a, b, c in combinations:
        if board[a] == board[b] == board[c] == p:
            return True

    return False


def minimax(maximizing):
    if win("O"):
        return 1
    if win("X"):
        return -1
    if " " not in board:
        return 0

    if maximizing:
        best = -100

        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(False)
                board[i] = " "
                best = max(best, score)

        return best

    else:
        best = 100

        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                score = minimax(True)
                board[i] = " "
                best = min(best, score)

        return best


def best_move():
    best_score = -100
    move = 0

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(False)
            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    return move


# Main game
while True:

    display()

    # Human move
    move = int(input("Enter your move (1-9): ")) - 1

    if board[move] != " ":
        print("Cell already occupied!")
        continue

    board[move] = "X"

    if win("X"):
        display()
        print("You win!")
        break

    if " " not in board:
        display()
        print("Draw!")
        break

    # Computer move
    move = best_move()
    board[move] = "O"

    print("Computer chose:", move + 1)

    if win("O"):
        display()
        print("Computer wins!")
        break
```
