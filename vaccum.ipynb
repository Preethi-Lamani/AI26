```python
A = input("Enter Room A (CLEAN/DIRTY): ").upper()
B = input("Enter Room B (CLEAN/DIRTY): ").upper()
pos = input("Enter Vacuum Position (A/B): ").upper()

if pos == "A":
    if A == "DIRTY":
        print("SUCK")
        A = "CLEAN"

    if B == "DIRTY":
        print("MOVE RIGHT")
        print("SUCK")
        B = "CLEAN"

else:
    if B == "DIRTY":
        print("SUCK")
        B = "CLEAN"

    if A == "DIRTY":
        print("MOVE LEFT")
        print("SUCK")
        A = "CLEAN"

print("Final State:")
print("Room A =", A)
print("Room B =", B)
```
