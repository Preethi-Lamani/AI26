```python
# Truth Table Enumeration Algorithm

# Implication function: P -> Q
def implies(p, q):
    return (not p) or q


# Knowledge Base
def check_kb(p, q, r):
    return (
        implies(q, p) and
        implies(p, not q) and
        (q or r)
    )


# Check whether KB entails a query
def tt_entails(query):
    symbols = [True, False]

    for p in symbols:
        for q in symbols:
            for r in symbols:

                kb = check_kb(p, q, r)

                if kb:
                    if not query(p, q, r):
                        return False

    return True


# Main program
print("TRUTH TABLE ENUMERATION")
print("-----------------------")

# Question 1: KB entails R?
result1 = tt_entails(lambda p, q, r: r)

# Question 2: KB entails R -> P?
result2 = tt_entails(lambda p, q, r: implies(r, p))

# Question 3: KB entails Q -> R?
result3 = tt_entails(lambda p, q, r: implies(q, r))


print("1. KB entails R:", result1)
print("2. KB entails R -> P:", result2)
print("3. KB entails Q -> R:", result3)
```

```python
def implies(a, b):
    return not a or b

Q = bool(int(input("Enter Q (0 or 1): ")))
P = bool(int(input("Enter P (0 or 1): ")))
R = bool(int(input("Enter R (0 or 1): ")))

kb = implies(Q, P) and implies(P, not Q) and (Q or R)

print("\nKnowledge Base:", kb)

if kb:
    print("KB is TRUE")

    query = int(input("\nSelect Query:\n1. R\n2. R -> P\n3. Q -> R\nEnter choice: "))

    if query == 1:
        result = R
    elif query == 2:
        result = implies(R, P)
    elif query == 3:
        result = implies(Q, R)
    else:
        print("Invalid choice")
        exit()

    print("Query is:", result)
    print("KB entails Query:", result)

else:
    print("KB is FALSE")
    print("This model cannot be used to check entailment.")
```
