```python
def unify(E1, E2):

    # Check predicate symbols
    if E1[0] != E2[0]:
        return "FAIL"

    # Check number of arguments
    if len(E1[1]) != len(E2[1]):
        return "FAIL"

    subst = {}

    # Compare arguments
    for a, b in zip(E1[1], E2[1]):

        if a.islower():
            if a in subst and subst[a] != b:
                return "FAIL"
            subst[a] = b

        elif b.islower():
            if b in subst and subst[b] != a:
                return "FAIL"
            subst[b] = a

        elif a != b:
            return "FAIL"

    return subst


# User input
p1 = input("Enter first predicate: ")
a1 = input("Enter arguments of first predicate: ").split(",")

p2 = input("Enter second predicate: ")
a2 = input("Enter arguments of second predicate: ").split(",")

E1 = (p1, a1)
E2 = (p2, a2)

result = unify(E1, E2)

print("Unification Result:", result)
```
