```python
def manhattan_distance(state, goal):
    h = 0

    for i in range(9):
        if state[i] != 0:

           
            row = i // 3
            col = i % 3

         
            goal_index = goal.index(state[i])
            goal_row = goal_index // 3
            goal_col = goal_index % 3

            h += abs(row - goal_row) + abs(col - goal_col)

    return h


state = [
    1, 2, 3,
    8, 6, 4,
    7, 5, 0
]


goal = [
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
]

print("Manhattan Distance =", manhattan_distance(state, goal))
```
